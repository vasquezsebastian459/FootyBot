```python
import os
import json


# Define the base directory relative to the current script location
base_path = os.path.join(os.getcwd(), '..', 'chat_responses')

performance_data_list = []  # List to store all performance data entries

# Iterate over each directory in the base path
for date_folder in os.listdir(base_path):
    
    date_folder_path = os.path.join(base_path, date_folder)
    
    # Check if the path is a directory
    if os.path.isdir(date_folder_path):
        # List all files in this directory
        for file_name in os.listdir(date_folder_path):
            
            if file_name.endswith('.json'):  # Ensure we are only reading JSON files                   
                file_path = os.path.join(date_folder_path, file_name)

                # Open and read the JSON file
                try:
                    with open(file_path, 'r') as file:
                        data = json.load(file)                            
                        
                        if data['Performance Data']:
                            file_path = os.path.join(date_folder_path, file_name)

                            data = {'file_path' : file_path, 'data' : data}
                            performance_data_list.append(data)                    
                except Exception as e:
                    print(f"Failed to read {file_name}: {str(e)}")


print(performance_data_list)
```

```python
import csv
import json
import pandas as pd
with open('data/performance_responses.json', 'r') as file:
    data = json.load(file)

# Extract performance data
performance_data = []
for item in data:
    performance_data.append(item["data"]["Performance Data"])

# Flatten performance data (recursive)
def flatten_dict(data, parent_key="", sep="."):
    items = []
    for key, value in data.items():
        new_key = parent_key + sep + key if parent_key else key
        if isinstance(value, dict):
            items.extend(flatten_dict(value, new_key).items())
        else:
            items.append((new_key, value))
    return dict(items)

# Flatten performance data
flat_performance_data = [flatten_dict(d) for d in performance_data]

# Get all keys from flattened performance data
keys = set.union(*[set(d.keys()) for d in flat_performance_data])

# Write data to CSV file
with open("performance_data.csv", "w", newline="") as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=keys)
    writer.writeheader()
    writer.writerows(flat_performance_data)

print("Performance data saved to performance_data.csv")
```

## Group all the questions and answers

```python
import os
import json


# Define the base directory relative to the current script location
base_path = os.path.join(os.getcwd(), '..', 'chat_responses')

question_list = []  # List to store all performance data entries

# Iterate over each directory in the base path
for date_folder in os.listdir(base_path):
    
    date_folder_path = os.path.join(base_path, date_folder)
    
    # Check if the path is a directory
    if os.path.isdir(date_folder_path):
        # List all files in this directory
        for file_name in os.listdir(date_folder_path):
            
            if file_name.endswith('.json'):  # Ensure we are only reading JSON files                   
                file_path = os.path.join(date_folder_path, file_name)

                # Open and read the JSON file
                try:
                    with open(file_path, 'r') as file:
                        data = json.load(file)                            
                        
                        if data['User Question']:
                            interaction = {
                                'User Question': data['User Question'],
                                'Generated Explanation': data['Generated Explanation']
                            }
                            question_list.append(interaction)                    
                except Exception as e:
                    file_path = os.path.join(date_folder_path, file_name)
                    print(f"Failed to read {file_path}: {str(e)}")




print(question_list)
```

```python
grouped_explanations = {}

for entry in question_list:
    question = entry['User Question']
    explanation = entry['Generated Explanation']
    
    # Check if the question already exists as a key in the dictionary
    if question in grouped_explanations:
        grouped_explanations[question].append(explanation)
    else:
            grouped_explanations[question] = [explanation]
```

```python
print(len(grouped_explanations))

with open('grouped_explanations.json', 'w') as file:
    json.dump(grouped_explanations, file, indent=4)
```
